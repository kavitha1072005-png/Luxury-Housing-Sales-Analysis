```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\HP\Downloads\Luxury_Housing_Bangalore.csv")
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Buyer_Comments"] = df["Buyer_Comments"].fillna("No Comment")
```

```python
df["Buyer_Comments"].isnull().sum()
```

```python
df["Unit_Size_Sqft"].median()
```

```python
df["Unit_Size_Sqft"] = df["Unit_Size_Sqft"].fillna(5990.0)
```

```python
df["Unit_Size_Sqft"].isnull().sum()
```

```python
df["Ticket_Price_Cr"].head(10)
```

```python
df["Ticket_Price_Cr"] = (
    df["Ticket_Price_Cr"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Cr", "", regex=False)
    .str.strip()
)
```

```python
df["Ticket_Price_Cr"] = pd.to_numeric(
    df["Ticket_Price_Cr"], errors="coerce"
)
```

```python
df["Ticket_Price_Cr"].dtype
```

```python
df["Ticket_Price_Cr"].isnull().sum()
```

```python
df["Ticket_Price_Cr"].median()
```

```python
df["Ticket_Price_Cr"] = df["Ticket_Price_Cr"].fillna(12.03764653480014)
```

```python
df["Ticket_Price_Cr"].isnull().sum()
```

```python
df["Amenity_Score"].median()
```

```python
df["Amenity_Score"] = df["Amenity_Score"].fillna(7.4991227969956995)
```

```python
df["Amenity_Score"].isnull().sum()
```

```python
df["Micro_Market"].unique()[:20]
```

```python
df["Micro_Market"] = df["Micro_Market"].str.strip().str.title()
```

```python
df["Micro_Market"].unique()[:20]
```

```python
df["Developer_Name"].unique()[:20]
```

```python
df["Price_per_Sqft"] = (df["Ticket_Price_Cr"] * 10000000) / df["Unit_Size_Sqft"]
```

```python
df[["Ticket_Price_Cr", "Unit_Size_Sqft", "Price_per_Sqft"]].head()
```

```python
df["Purchase_Quarter"].unique()
```

```python
df["Purchase_Quarter"] = pd.to_datetime(df["Purchase_Quarter"])
```

```python
df["Quarter_Number"] = df["Purchase_Quarter"].dt.quarter
```

```python
df[["Purchase_Quarter", "Quarter_Number"]].head(10)
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.to_csv("Luxury_Housing_Bangalore_Cleaned.csv", index=False)
```
